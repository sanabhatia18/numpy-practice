2. Generate an array of 10 random integers between 1 and 100 and find their mean, median, and standard deviation.

```python
a=np.random.randint(1,100,10)
print(a)
print(a.mean())  #mean
print(np.quantile(a,0.5))  #median
print(np.median(a))    #median  
print(np.std(a))   #standard dev
```

3. Create two arrays [1,2,3] and [4,5,6] and perform element-wise addition, subtraction, multiplication, and division.

```python
ar1=np.array([1,2,3])
print(ar1)
print("\n")

ar2=np.array([4,5,6])
print(ar2)
print("\n")

print(ar1+ar2) #add
print(ar1-ar2)  #subtract
print(ar1*ar2)   #multiply
print(ar1/ar2)  #divide
```

4. Given an array [10, 20, 30, 40, 50], use np.where to label values greater than 25 as 'High' and others as 'Low'.

```python
ar=np.array([10,20,30,40,50])
print(ar)
a=ar[ar>25]
print(a)
b=ar[ar<25]
print(b)
print(np.where(ar>25,"high","low"))
```

6. Create an array of 15 random numbers and find the maximum and minimum values with their indices.

```python
arr=np.random.randint(1,100,15)
print(arr)

print("indices",arr.argsort())
print("sorted array" ,arr[np.argsort(arr)])
print("max value", max(arr))
print("min value",min(arr))
```

8. Create an array of 100 random numbers and find the top 5 maximum values using np.argsort.

```python
arr=np.random.randint(1,200,100)
print("array = " , arr)
print("\n")

a=arr[np.argsort(arr)]
print("sorted = ",a)
print("\n")

print("top 5 max" , a[100:94:-1])
```
