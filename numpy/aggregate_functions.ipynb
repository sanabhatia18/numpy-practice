```python
import numpy as np
```

```python
#aggregate functions
ar=np.array(([11,22,33],[100,100,100],[7,8,9]))
print(ar)
ar=np.arange(1,10).reshape(3,3)
print(ar)
```

```python
#broadcasting
a1=np.arange(1,9)
print(a1)
a2=np.arange(11,99,11)
print(a2)
print(a1+a2)
print(a1*a2)
print(a1.min())
print(a1.max())
print(a1.mean())
print(len(a1))
print(np.quantile(a2,0.5))

La1=[1,2,3,4,5,6,7,8]
la2=[11,22,33,44,55,66,77,88]
print(La1+la2) #it concatenates
```

```python
print(ar)
print(ar<5)
print(ar[ar<5])
ar=np.random.randint(1,100,50)
print(ar)
print(ar[(ar<50)&(ar<80)])
a1=np.random.randint(1,100,9).reshape(3,3)
a2=np.random.randint(1,100,9).reshape(3,3)
print(a1)
print(a2)
print(np.concatenate([a1,a2],axis=0)) #row concatenate
print(np.concatenate([a1,a2],axis=1))  #col concatenate
print(np.vdot(a1,a2)) #mult ka sum
print((a1*a2).sum())  #same as above

```

```python
#ravel and flatten
print(a1)
fa=a1.flatten()  #generates duplicate copies also
print(fa)
fa[0]=100
print(fa)
print("\n")

print(a2)
ra=a2.ravel() #generates REFERENCE COPY IN 1S ARRAy
ra[0]=100
print(ra)

print("\n")

print(np.split(ra,3))  #splits the array
rav=np.vstack(ra)   #vertically stacks the array
print(np.hstack(rav))  #horizonatally
print(np.histogram(ra,bins=3))
print("\n")

print(ra.sort())
A=np.random.randint(1,100,50)
print(np.argsort(A))  #index deta 
print(A[np.argsort(A)])  #ascending
```
