```python
import pandas as pd
Path=r'C:\Users\sanab\Desktop\pandas\Superstore.xlsx'
Path
orders=pd.read_excel(Path)
orders
df=pd.pivot_table(orders,index='Region',columns='Category',values='Sales',aggfunc='sum').reset_index()
df
```

```python
pd.melt(df,id_vars='Region').rename(columns={'value':'Sales'})
```
