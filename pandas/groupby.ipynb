```python
orders=pd.read_excel(Path)
orders
```

# GROUP BY

```python
orders.groupby('Region').agg({'Sales' : 'sum'})
```

```python
#multiple aggregations
orders.groupby('Region').agg({'Sales':'sum','Profit':'sum'})  
```

```python
#multiple group by
orders.groupby(['Region','State']).agg({'Sales':'sum','Profit':'sum'})
```

```python
#find out top 3 state having highest profit
orders.groupby('State').agg({'Profit':'sum'}).sort_values('Profit',ascending=False).head(3)
```

# CROSS TABULATION

```python
pd.crosstab(orders['Region'],orders['Category'])
```

# JOINS

```python
cust=pd.DataFrame({'cust_id':[1,2,3,4,5],'cust_name':['raam','shyam','hari','murli','krishna']})
cust
```

```python
bill=pd.DataFrame({'order_id':[11,22,33,44,55],'cust_id':[1,1,2,3,6],'product':['lap','mouse','keyboard','console','monitor']})
bill
```

```python
#matching from both table

pd.merge(cust,bill,on='cust_id',how='inner')
```

```python
#all from right table and matching from right table

pd.merge(cust,bill,on='cust_id',how='left')
```

```python
#all from right table and matching from left table

pd.merge(cust,bill,on='cust_id',how='right') 
```

```python
#all data from both table

pd.merge(cust,bill,on='cust_id',how='outer')
```
